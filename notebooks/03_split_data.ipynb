```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. قراءة البيانات من نوتبوك 2
df = pd.read_parquet("../artifacts/ml_orders_with_labels.parquet")

# 2. التقسيم الأول: Train (70%) و Temp (30%)
train_df, temp_df = train_test_split(
    df, 
    test_size=0.30, 
    random_state=42, 
    stratify=df["is_late"]
)

# 3. التقسيم الثاني: Val (15%) و Test (15%)
val_df, test_df = train_test_split(
    temp_df, 
    test_size=0.50, 
    random_state=42, 
    stratify=temp_df["is_late"]
)

# 4. طباعة النتائج والتأكد من التوازن
print(f"Train set: {train_df.shape}, Late Ratio: {train_df['is_late'].mean():.4f}")
print(f"Val set:   {val_df.shape}, Late Ratio: {val_df['is_late'].mean():.4f}")
print(f"Test set:  {test_df.shape}, Late Ratio: {test_df['is_late'].mean():.4f}")

# 5. حفظ الأجزاء الثلاثة كـ Artifacts
train_df.to_parquet("../artifacts/train.parquet", index=False)
val_df.to_parquet("../artifacts/val.parquet", index=False)
test_df.to_parquet("../artifacts/test.parquet", index=False)

print("تم الحفظ بنجاح!")
```
