```python
import pandas as pd

# 1. قراءة الـ Artifact من الخطوة السابقة
df = pd.read_parquet("../artifacts/ml_orders_table.parquet")

# 2. تحويل أعمدة التواريخ لـ datetime
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_cols:
    df[col] = pd.to_datetime(df[col])

# 3. إنشاء الـ Target Variable (هل الشحنة تأخرت عن الموعد المتوقع؟)
# is_late = 1 إذا كان تاريخ التسليم الفعلي بعد التاريخ المتوقع
df["is_late"] = (df["order_delivered_customer_date"] > df["order_estimated_delivery_date"]).astype(int)

# 4. طباعة نسبة التأخير للتحقق من التوازن (Class Balance)
print("توزيع الـ Target Variable (is_late):")
print(df["is_late"].value_counts(normalize=True))

# 5. حفظ الـ Artifact الجديد
output_path = "../artifacts/ml_orders_with_labels.parquet"
df.to_parquet(output_path, index=False)
print(f"تم حفظ الـ Artifact بنجاح في: {output_path}")
```
